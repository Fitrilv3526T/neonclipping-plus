### Troubleshooting

- **The tunnel link does not open** - re-run the setup cell; it is idempotent and will
  restart only what is missing.
- **YouTube says the server is blocked** - upload a cookies.txt in the app's Settings,
  or upload the MP4 directly. Colab IPs are blocked often; this is not fixable in code.
- **No GPU** - everything still runs, captions are slower (small/base Whisper on CPU).
- **Everything disappeared** - Colab free sessions are temporary. Save your clips to
  Drive (cell above) before the session ends.
